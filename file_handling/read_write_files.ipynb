File Operation: Read and Write Files

```python
#Read a file line by line
with open("example.txt", "r") as file:
    for line in file:
        print(line.strip())
        # .strip() removes any leading/trailing whitespace characters, including the newline character at the end of each line.
```

```python
#Writing a file/ overwriting a file
with open("output.txt", "w") as file:
    file.write("This is a line of text.\n")
    file.write("This is another line of text.\n")
```

```python
#Write a file without overwriting
with open("example.txt", "a") as file:
    file.write("This line will be appended to the file.\n")
#Appending multiple lines
lines_to_append = ["First appended line.\n", "Second appended line.\n", "Third appended line.\n"]
with open("example.txt", "a") as file:
    file.writelines(lines_to_append)
    
```

```python
#Binary files
#Writing to a binary file

data = b'\x00\x01\x02\x03\x04'
with open("binary.bin", "wb") as file:
    file.write(data)
#Reading from a binary file
with open("binary.bin", "rb") as file:
    binary_data = file.read()
    print(binary_data)  
```

```python
##Read the content from source text file and write it to a dest text file
with open("example.txt", "r") as source_file:
    content = source_file.read()
with open("dest.txt", "w") as dest_file:
    dest_file.write(content)
    
```

```python
#Writing and then reading a file

with open("example.txt", "w+") as file:
    file.write("w+ mode allows both writing and reading.\n")
    file.write("You can write to the file and then read from it without closing it.\n")
    #seek is used to move the file pointer to a specific position in the file.
    file.seek(0)        
    content = file.read()
    print(content)
    
```

```python
#create new directory and write a file in that directory
import os
os.makedirs("new_directory", exist_ok=True)
with open("new_directory/new_file.txt", "w") as file:
    file.write("This file is created inside a new directory.\n")    
    
```

```python
#listing files and dirs
items = os.listdir(".")
print(items)
```

```python
#joining paths(imp feature of os)
dir_name = "folder"
file_name = "file.txt"
full_path = os.path.join(dir_name, file_name)
print(full_path)
```

```python
#joining paths(imp feature of os)
dir_name = "folder"
file_name = "file.txt"
full_path = os.path.join(os.getcwd(), dir_name, file_name)
print(full_path)
```
