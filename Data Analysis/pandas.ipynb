Pandas - DataFrame and Series

```python
import pandas as pd
data = [1,2,3,4,5]
series = pd.Series(data)
print(f"Series::::::{series}")
```

```python
# Create Series from dict

data = {'a':1, "b":2}
series_dict = pd.Series(data)
print(f"data::::{series_dict}")
```

```python
data = [10,20,30]
index = ['a', 'b', 'c']
pd.Series(data, index=index)
```

```python
# Dataframe
# create a df from a dict of list

data = {
    'Name': ['Swag', 'John', 'Jack'],
    'Age': [25, 30, 45],
    'City': ['Bhubaneswar', 'New York', 'Chicago']
}

df = pd.DataFrame(data)
print(df)


```

```python
import numpy as np
np.array(df)
```

```python
# Create a df from list of dict
data = [
    {'Name': 'Swag', 'Age': 32, 'City': 'Bhubaneswar'},
    {'Name': 'John', 'Age': 34, 'City': 'Bhubaneswar'},
    {'Name': 'Jack', 'Age': 32, 'City': 'Bhubaneswar'},
    {'Name': 'Sway', 'Age': 32, 'City': 'Bhubaneswar'},
]
df = pd.DataFrame(data)
print(df)
```

```python
# Data manipulation with df
# Add a new col
df['Salary'] = [1000000, 900000, 800000, 800000]
df
# Remove a col
df.drop('Salary', axis=1, inplace=True)
df
```

```python
# Add age to col
df['Age'] = df['Age'] + 1
```
